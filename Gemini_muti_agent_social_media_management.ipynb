```python
import os
import nest_asyncio
import requests
from google.colab import userdata
from crewai import Agent, Task, Crew, LLM
from crewai.tools import tool # Import pour créer des outils
from IPython.display import Image, display # Pour afficher l'image dans Colab

# 1. Configuration de base
nest_asyncio.apply()
os.environ["GEMINI_API_KEY"] = userdata.get('GEMINI_API_KEY')

cerveau_gemini = LLM(
    model="gemini/gemini-3.5-flash-lite",
    api_key=os.environ["GEMINI_API_KEY"]
)

# ==========================================
# 2. NOUVEAU : CRÉATION DE L'OUTIL POUR L'AGENT
# ==========================================
@tool("Générateur d'images gratuit")
def generer_image(prompt_anglais: str) -> str:
    """
    Génère une image à partir d'une description textuelle (prompt) en ANGLAIS et la sauvegarde.
    """
    print(f"\n🎨 [L'Agent dessine] -> {prompt_anglais}")
    # Appel à une API gratuite pour générer l'image
    url = f"https://image.pollinations.ai/prompt/{prompt_anglais.replace(' ', '%20')}?width=1024&height=1024&nologo=true"
    reponse = requests.get(url)

    nom_fichier = "image_generale_post.jpg"
    with open(nom_fichier, 'wb') as fichier:
        fichier.write(reponse.content)

    return f"Image générée avec succès et sauvegardée sous le nom : {nom_fichier}"

# ==========================================
# 3. TES AGENTS (Mise à jour du Designer)
# ==========================================
createur_contenu = Agent(
    role='Créateur de Contenu Viral',
    goal='Créer des textes engageants et captivants à partir d\'une idée de base.',
    backstory='Tu es un expert en marketing digital avec une plume créative.',
    llm=cerveau_gemini,
    verbose=True,
    allow_delegation=False
)

designer_visuel = Agent(
    role='Designer Visuel et Graphiste IA',
    goal='Traduire le sujet du post en une courte description en anglais et GÉNÉRER l\'image avec ton outil.',
    backstory='Tu es un Directeur Artistique. Tu sais résumer un texte en un prompt visuel en anglais, puis tu utilises ton Générateur d\'images pour créer le fichier.',
    llm=cerveau_gemini,
    tools=[generer_image], # <--- On donne l'outil à l'agent ici !
    verbose=True,
    allow_delegation=False
)

social_media_manager = Agent(
    role='Social Media Manager Senior',
    goal='Finaliser les textes et confirmer que l\'image est prête.',
    backstory='Tu es un stratège des réseaux sociaux. Tu prépares les versions finales des textes.',
    llm=cerveau_gemini,
    verbose=True,
    allow_delegation=False
)

# ==========================================
# 4. TES TÂCHES (Mise à jour de la tâche de design)
# ==========================================
redaction_task = Task(
    description= input("entrez le sujet:"),
    expected_output='Un brouillon de post de 2 ou 3 paragraphes maximum.',
    agent=createur_contenu
)

design_task = Task(
    description='1. Lis le post rédigé.\n2. Invente un prompt visuel court EN ANGLAIS décrivant une image d\'illustration.\n3. UTILISE obligatoirement l\'outil "Générateur d\'images gratuit" avec ton prompt en anglais pour fabriquer l\'image.',
    expected_output='Le message de confirmation que l\'image a été créée (ex: image_generale_post.jpg).',
    agent=designer_visuel
)

optimisation_task = Task(
    description='Prends le brouillon du créateur, crée une version Facebook et une version LinkedIn prêtes à publier.',
    expected_output='Les posts finaux avec émojis et hashtags.',
    agent=social_media_manager
)

# 5. Lancement de l'équipe
mon_equipe_social_media = Crew(
    agents=[createur_contenu, designer_visuel, social_media_manager],
    tasks=[redaction_task, design_task, optimisation_task],
    verbose=True
)

print("🚀 L'équipe se met au travail. Le designer va générer un vrai fichier image...")
resultat_final = await mon_equipe_social_media.kickoff_async()

print("\n\n========================================")
print("🎉 TEXTES FINAUX :")
print("========================================\n")
print(resultat_final)

# 6. Affichage de l'image générée directement dans Colab
print("\n\n🖼️ VOICI L'IMAGE GÉNÉRÉE PAR TON AGENT :")
try:
    display(Image(filename='image_generale_post.jpg'))
except Exception as e:
    print("L'image n'a pas pu être affichée, vérifie les dossiers à gauche.")
```
