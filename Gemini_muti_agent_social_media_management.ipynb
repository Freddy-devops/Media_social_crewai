```python
import os
import nest_asyncio
from google.colab import userdata
from crewai import Agent, Task, Crew, LLM

# 1. On règle le conflit Colab / CrewAI
nest_asyncio.apply()

# 2. On récupère ta clé Gemini
os.environ["GEMINI_API_KEY"] = userdata.get('GEMINI_API_KEY')

# 3. LE CORRECTIF EST ICI : on change le nom du modèle pour "gemini-pro"
cerveau_gemini = LLM(
    model="gemini/gemini-3.5-flash-lite",
    api_key=os.environ["GEMINI_API_KEY"]
)

# 4. Tes Agents
createur_contenu = Agent(
    role='Créateur de Contenu Viral',
    goal='Créer des textes engageants et captivants à partir d\'une idée de base.',
    backstory='Tu es un expert en marketing digital avec une plume créative, capable de transformer n\'importe quel sujet en une histoire passionnante.',
    llm=cerveau_gemini,
    verbose=True,
    allow_delegation=False
)

social_media_manager = Agent(
    role='Social Media Manager Senior',
    goal='Optimiser le contenu pour maximiser l\'engagement. Ajuster le ton pour LinkedIn et Facebook.',
    backstory='Tu es un stratège des réseaux sociaux. Tu sais exactement quels hashtags utiliser et comment formater un post.',
    llm=cerveau_gemini,
    verbose=True,
    allow_delegation=True
)

# 5. Tes Tâches
redaction_task = Task(
    description='Rédige un post sur l\'importance de l\'intelligence artificielle pour automatiser ses tâches quotidiennes. Le texte doit être clair, inspirant et personnel.',
    expected_output='Un brouillon de post de 2 ou 3 paragraphes maximum.',
    agent=createur_contenu
)

optimisation_task = Task(
    description='Prends le brouillon du créateur et crée deux versions finales : une version avec des émojis et hashtags pour Facebook, et une version plus professionnelle et structurée pour LinkedIn.',
    expected_output='Deux posts distincts prêts à être publiés : 1 pour Facebook, 1 pour LinkedIn.',
    agent=social_media_manager
)

# 6. L'Équipe et le Lancement
mon_equipe_social_media = Crew(
    agents=[createur_contenu, social_media_manager],
    tasks=[redaction_task, optimisation_task],
    verbose=True
)

print("🚀 L'équipe se met au travail avec Gemini Pro... (Patiente un instant)")
resultat_final = await mon_equipe_social_media.kickoff_async()

print("\n\n========================================")
print("🎉 VOICI TES POSTS PRÊTS À ÊTRE PUBLIÉS :")
print("========================================\n")
print(resultat_final)
```
